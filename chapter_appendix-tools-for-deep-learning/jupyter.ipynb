```python
def find_pattern(text,pattern):    # creating a userdefined function for resuibility
  if re.search(pattern,text):
    return re.search(pattern,text)
  else:
    return 'Not found'
#
```

Quantifiers




In Regular Expressions (Regex), quantifiers define how many times a character, group, or token should appear in the matching pattern.

They control the repetition of elements in the pattern.

```python
Quantifiers *'?': Zero or one (tells whether a pattern is absent or present) print(find_pattern("ac", "ab?")) print(find_pattern("abc", "ab?")) print(find_pattern("abbc", "ab?")) [] +': One or more print(find_pattern("ac", "ab+")) print(find_pattern("abc", "ab+")) print(find_pattern("abbc", "ab+")) [] (n): Matches if a character is present exactly n number of times print(find_pattern("abbc", "ab{2}"))
```

```python
# "?" : zero or one ("tells whether a pattern is absent or present ")

print(find_pattern('ac','ab?'))   # only a found
print(find_pattern('abc','ab?'))  # both
print(find_pattern('abbc','ab?'))    # text , pattern
```

in the above example

 we have pattern  ab ? it is trying to fing out the "b" just after
  "a"



 String: "abbc"
Pattern: "ab?"

The first 'ab' is matched

The second 'b' is not included because ? allows only 0 or 1 b

```python
# "*" : zero or more ("tells whether a pattern is absent or present ")

print(find_pattern('ac','ab*'))   # only a found
print(find_pattern('abc','ab*'))  # both
print(find_pattern('abbc','ab*'))    # text , pattern    mutliple times
```

```python
# "+" one or more
print(find_pattern('ac','ab+'))   # only a found
print(find_pattern('abc','ab+'))
print(find_pattern('abbc','ab+'))
```

```python
# {'n'} : Matches if a character is present exactly  n muber of times
print(find_pattern('abbbc','ab{2}'))  #  n=2
print(find_pattern('abbbc','ab{4}'))  #  n=4
```

```python
#{m,n} :  Matches if a character is present  from m to n number of times
#  here m is starting position and n is ending position


print(find_pattern('aabbbbbbc','ab{3,5}'))   # retrun  true if 'b' is present 3-5 times
print(find_pattern('aabbbbbbc','ab{7,10}')) #7-10 times
print(find_pattern('aabbbbbbc','ab{,10}'))  # almost 10 times
print(find_pattern('aabbbbbbc','ab{10,}')) # at least 10 times
#
```
